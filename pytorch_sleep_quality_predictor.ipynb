```python
import torch
import torch.nn as nn
import torch.optim as optim

print("SLEEP QUALITY PREDICTOR")

screen_time = torch.tensor(
    [[1], [2], [3], [4], [5], [6]],
    dtype=torch.float32
)

sleep_quality = torch.tensor(
    [[95], [90], [85], [75], [65], [50]],
    dtype=torch.float32
)

model = nn.Linear(1, 1)

criterion = nn.MSELoss()

optimizer = optim.SGD(
    model.parameters(),
    lr=0.01
)

for epoch in range(500):

    prediction = model(screen_time)

    loss = criterion(
        prediction,
        sleep_quality
    )

    optimizer.zero_grad()

    loss.backward()

    optimizer.step()

prediction = model(
    torch.tensor([[7.0]])
)

print(
    f"Predicted Sleep Quality for 7 hours screen time: "
    f"{prediction.item():.2f}"
)
```
